```python
import urllib
import datetime
import requests
import csv
import pandas as pd


url = 'https://api.edinet-fsa.go.jp/api/v2/documents.json'

key = '6f25302ddba748ba866ae2164b963e02'

# 设置测试的时间范围（一周）
end_date = datetime.date.today()
start_date = datetime.date(2020, 3, 1)

# 创建空的DataFrame来存储结果
all_results = pd.DataFrame()

# 按日循环获取数据
current_date = start_date
while current_date <= end_date:
    # 构建API请求参数
    params = {
        'date': current_date.strftime('%Y-%m-%d'),
        'type': 2,  # 取得書類種別（2: 提出書類）
        'Subscription-Key': key
    }
    
    try:
        # 调用API
        response = requests.get(url, params=params)
        data = response.json()
        #print(data)
        # 如果当天有数据，将其添加到DataFrame中
        if data['metadata']['status'] == '200' and data['metadata']['resultset']['count'] > 0:
            df = pd.DataFrame(data['results'])
            df['date'] = current_date
            all_results = pd.concat([all_results, df], ignore_index=True)
            print(f"成功获取 {current_date} 的数据，数量: {len(df)}")
        else:
            print(f"{current_date} 没有可用数据")
            
    except Exception as e:
        print(f"获取 {current_date} 的数据时发生错误: {str(e)}")
    
    # 移到下一天
    current_date += datetime.timedelta(days=1)

# 显示结果统计
print("\n数据获取完成！")
print(f"总记录数: {len(all_results)}")
print("\n数据框的前几行:")
display(all_results.head())
```

```python
import os
import datetime
import requests
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry
import time
# 下载Xbrl
base_url = 'https://disclosure.edinet-fsa.go.jp/api/v2/documents/'
download_path = '/Users/wenjun/Downloads/JPXData/Yuho'
os.makedirs(download_path, exist_ok=True)

def get_existing_files():
    """获取已下载文件的列表"""
    existing_files = set()
    for filename in os.listdir(download_path):
        if filename.endswith('.zip'):
            # 从文件名中提取 secCode 和 submitDate
            parts = filename.replace('.zip', '').split('_')
            if len(parts) == 2:
                existing_files.add(tuple(parts))
    return existing_files


# 设置重试策略
retry_strategy = Retry(
    total=20,  # 最大重试次数
    backoff_factor=2,  # 重试间隔时间（会按2, 4, 8秒递增）
    status_forcelist=[429, 500, 502, 503, 504],  # 需要重试的HTTP状态码
)

# 创建会话对象
session = requests.Session()
adapter = HTTPAdapter(max_retries=retry_strategy)
session.mount("https://", adapter)


def download_xbrl(doc_id, save_path, secCode, submitDate):
    try:# EDINET API endpoint
    #submitDate = submitDate.strftime('%Y-%m-%d %H:%M')
        submitDate = datetime.datetime.strptime(submitDate,'%Y-%m-%d %H:%M').strftime('%Y%m%d')
        url = f'{base_url}{doc_id}'
        
        # API参数
        params = {
            'type': 1,  # 1 表示 XBRL
            'Subscription-Key': key
        }
        # 添加延迟，避免请求过于频繁
        time.sleep(2)
        response = session.get(url, params=params, stream=True, timeout=30)
        response.raise_for_status()  # 如果状态码不是200，会抛出异常
        
        file_path = os.path.join(save_path, f'{secCode}_{submitDate}.zip')
        with open(file_path, 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                if chunk:
                    f.write(chunk)
        print(f"成功下载文件到: {file_path}")
        return True
        
    except requests.exceptions.RequestException as e:
        print(f"下载失败 {doc_id}: {str(e)}")
        return False
    except Exception as e:
        print(f"发生其他错误 {doc_id}: {str(e)}")
        return False

#download_xbrl('S100SKCP', download_path, '36910', '2024-01-04 12:01')
```

```python
def batch_download(filtered_df):
    # 设置重试策略
    retry_strategy = Retry(
        total=3,
        backoff_factor=2,
        status_forcelist=[429, 500, 502, 503, 504],
    )
    
    session = requests.Session()
    adapter = HTTPAdapter(max_retries=retry_strategy)
    session.mount("https://", adapter)
    
    # 获取已下载文件列表
    existing_files = get_existing_files()
    
    success_count = 0
    fail_count = 0
    skip_count = 0
    
    for index, row in filtered_df.iterrows():
        # 转换日期格式
        submit_date = datetime.datetime.strptime(
            row['submitDateTime'],
            '%Y-%m-%d %H:%M'
        ).strftime('%Y%m%d')
        
        # 检查文件是否已存在
        file_key = (str(row['secCode']), submit_date)
        if file_key in existing_files:
            print(f"跳过已下载文件: {row['secCode']}_{submit_date}")
            skip_count += 1
            continue
            
        try:
            # 添加延迟
            time.sleep(2)
            
            url = f'{base_url}{row["docID"]}'
            params = {
                'type': 1,
                'Subscription-Key': key
            }
            
            response = session.get(url, params=params, stream=True, timeout=30)
            response.raise_for_status()
            
            file_path = os.path.join(download_path, f'{row["secCode"]}_{submit_date}.zip')
            with open(file_path, 'wb') as f:
                for chunk in response.iter_content(chunk_size=8192):
                    if chunk:
                        f.write(chunk)
                        
            success_count += 1
            print(f"成功下载文件到: {file_path}")
            
        except requests.exceptions.RequestException as e:
            print(f"下载失败 {row['docID']}: {str(e)}")
            fail_count += 1
        except Exception as e:
            print(f"发生其他错误 {row['docID']}: {str(e)}")
            fail_count += 1
            
        print(f"进度: 成功 {success_count}, 失败 {fail_count}, 跳过 {skip_count}")
    
    print(f"下载完成: 总成功 {success_count}, 总失败 {fail_count}, 总跳过 {skip_count}")
```

```python
# Extract XBRL files
import os
import zipfile
import glob
import xml.etree.ElementTree as ET
import tempfile
import shutil
from pathlib import Path


def extract_specific_xbrl(zip_path, output_dir):
    # 确保输出目录存在
    os.makedirs(output_dir, exist_ok=True)
    
    def extract_xbrl_info(file_path):
        tree = ET.parse(file_path)
        root = tree.getroot()
        
        result = {}
        for elem in root.iter():
            if elem.tag.endswith('FilingDateCoverPage'):
                result['filing_date'] = elem.text
            elif elem.tag.endswith('SecurityCodeDEI'):
                result['security_code'] = elem.text
            elif elem.tag.endswith('EDINETCodeDEI'):
                result['edinet_code'] = elem.text
        
        return result
    
    try:
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            # 获取所有文件列表
            file_list = zip_ref.namelist()
            #print(file_list)
            # 筛选符合条件的文件
            target_files = [f for f in file_list 
                          if ('XBRL/PublicDoc/' in f 
                          and f.startswith('XBRL/PublicDoc/jpcrp030000') 
                          and f.endswith('.xbrl')
                          and '_20' in f)]
            
            if target_files:
                print(f"在 {os.path.basename(zip_path)} 中找到符合条件的文件：")
                for file in target_files:
                    print(f"- {file}")
                    # 提取文件到输出目录
                    temp_dir = tempfile.mkdtemp()
                    zip_ref.extract(file, temp_dir)
                    temp_file_path = os.path.join(temp_dir, file)
                    #rint(f"  已保存到: {os.path.join(output_dir, file)}")
                    
                    try:
                        # 提取XBRL信息
                        info = extract_xbrl_info(temp_file_path)
                        # 构建新文件名
                        new_filename = f"{info['security_code']}_{info['filing_date']}_{info['edinet_code']}.xbrl"
                        new_file_path = os.path.join(output_dir, new_filename)
                        # 移动并重命名文件
                        os.makedirs(os.path.dirname(new_file_path), exist_ok=True)
                        os.rename(temp_file_path, new_file_path)
                        print(f"  已重命名并保存到: {new_file_path}")
                        
                    except Exception as e:
                        print(f"  处理文件时发生错误: {str(e)}")
                        # 如果处理失败，使用原始文件名
                        original_path = os.path.join(output_dir, file)
                        os.makedirs(os.path.dirname(original_path), exist_ok=True)
                        os.rename(temp_file_path, original_path)
                        print(f"  已使用原始文件名保存到: {original_path}")
                    
                    finally:
                        # 清理临时目录
                        import shutil
                        shutil.rmtree(temp_dir)
                return target_files
            else:
                print(f"在 {os.path.basename(zip_path)} 中未找到符合条件的文件")
                return []
    except Exception as e:
        print(f"处理 {zip_path} 时发生错误: {str(e)}")
        return []

# 主程序
yuho_path = '/Users/wenjun/Downloads/JPXData/Yuho'
output_dir = os.path.join(yuho_path, 'XBRL/')
zip_files = glob.glob(os.path.join(yuho_path, '*.zip'))

# 只处理前5个zip文件
print(f"开始处理，提取的文件将保存到: {output_dir}")
# for zip_file in zip_files[:5]:
for zip_file in zip_files:
    print("\n处理文件:", os.path.basename(zip_file))
    matching_files = extract_specific_xbrl(zip_file, output_dir)
```
